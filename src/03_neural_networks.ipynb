```python
print(data.shape)
print(data.columns.tolist())
print(data.head())
```

```python
features = [
    "hour",
    "day",
    "month",
    "day_of_week",
    "is_weekend",
    "lag_1",
    "lag_4",
    "lag_96",
    "lag_672",
    "rolling_mean_4",
    "rolling_mean_96"
]

target = "target_6h"

X = data[features]
y = data[target]

print("Features shape:", X.shape)
print("Target shape:", y.shape)
```

```python
# Calculate split positions
n = len(data)

train_end = int(n * 0.70)
val_end = int(n * 0.85)

# 6-hour horizon = 24 rows (15-minute intervals)
gap = 24

# Create datasets
train = data.iloc[:train_end].copy()

val = data.iloc[train_end + gap:val_end].copy()

test = data.iloc[val_end + gap:].copy()

print("Training:", train.shape)
print("Validation:", val.shape)
print("Testing:", test.shape)

print("\nDate ranges:")
print("Training:", train["date"].min(), "to", train["date"].max())
print("Validation:", val["date"].min(), "to", val["date"].max())
print("Testing:", test["date"].min(), "to", test["date"].max())
```

```python
from sklearn.preprocessing import StandardScaler

# Select features and target for each dataset
X_train = train[features]
X_val = val[features]
X_test = test[features]

y_train = train[target]
y_val = val[target]
y_test = test[target]

# Create scaler
scaler = StandardScaler()

# Fit only on training data
X_train_scaled = scaler.fit_transform(X_train)

# Apply the same scaling to validation and test
X_val_scaled = scaler.transform(X_val)
X_test_scaled = scaler.transform(X_test)

print("Training shape:", X_train_scaled.shape)
print("Validation shape:", X_val_scaled.shape)
print("Testing shape:", X_test_scaled.shape)
```

```python
import torch

# Convert features
X_train_tensor = torch.tensor(X_train_scaled, dtype=torch.float32)
X_val_tensor = torch.tensor(X_val_scaled, dtype=torch.float32)
X_test_tensor = torch.tensor(X_test_scaled, dtype=torch.float32)

# Convert targets
y_train_tensor = torch.tensor(
    y_train.values, dtype=torch.float32
).reshape(-1, 1)

y_val_tensor = torch.tensor(
    y_val.values, dtype=torch.float32
).reshape(-1, 1)

y_test_tensor = torch.tensor(
    y_test.values, dtype=torch.float32
).reshape(-1, 1)

print("X train:", X_train_tensor.shape)
print("y train:", y_train_tensor.shape)
```

```python
from torch.utils.data import TensorDataset, DataLoader

# Combine features and targets
train_dataset = TensorDataset(
    X_train_tensor,
    y_train_tensor
)

# Create batches
train_loader = DataLoader(
    train_dataset,
    batch_size=64,
    shuffle=True
)

print("Total training rows:", len(train_dataset))
print("Batch size:", 64)
print("Number of batches:", len(train_loader))
```

```python
import torch.nn as nn

class EnergyNN(nn.Module):

    def __init__(self):
        super().__init__()

        self.network = nn.Sequential(
            nn.Linear(11, 64),
            nn.ReLU(),

            nn.Linear(64, 32),
            nn.ReLU(),

            nn.Linear(32, 1)
        )

    def forward(self, x):
        return self.network(x)


model = EnergyNN()

print(model)
```

```python
# Define loss function
loss_function = nn.MSELoss()

# Define optimizer
optimizer = torch.optim.Adam(
    model.parameters(),
    lr=0.001
)

print("Loss function:", loss_function)
print("Optimizer:", optimizer)
```

```python
epochs = 50

for epoch in range(epochs):

    model.train()
    total_loss = 0

    for X_batch, y_batch in train_loader:

        # 1. Forward propagation
        predictions = model(X_batch)

        # 2. Calculate loss
        loss = loss_function(predictions, y_batch)

        # 3. Backpropagation
        optimizer.zero_grad()
        loss.backward()

        # 4. Update weights
        optimizer.step()

        total_loss += loss.item()

    average_loss = total_loss / len(train_loader)

    if (epoch + 1) % 5 == 0:
        print(
            f"Epoch [{epoch+1}/{epochs}], "
            f"Training Loss: {average_loss:.4f}"
        )
```

```python
model.eval()

with torch.no_grad():
    val_predictions = model(X_val_tensor)

    val_loss = loss_function(
        val_predictions,
        y_val_tensor
    )

print("Validation MSE:", val_loss.item())
print("Validation RMSE:", val_loss.item() ** 0.5)
```

```python
from sklearn.metrics import mean_absolute_error, mean_squared_error

model.eval()

with torch.no_grad():
    val_predictions = model(X_val_tensor).numpy()

# Convert actual values to NumPy
actual_values = y_val_tensor.numpy()

val_mae = mean_absolute_error(actual_values, val_predictions)
val_rmse = mean_squared_error(
    actual_values, val_predictions
) ** 0.5

print("Validation MAE:", round(val_mae, 4))
print("Validation RMSE:", round(val_rmse, 4))
```

```python
import copy

# Start with a fresh model
model = EnergyNN()

loss_function = nn.MSELoss()

optimizer = torch.optim.Adam(
    model.parameters(),
    lr=0.001
)

epochs = 150
patience = 15

best_val_loss = float("inf")
best_weights = None
epochs_without_improvement = 0

for epoch in range(epochs):

    model.train()
    total_train_loss = 0

    for X_batch, y_batch in train_loader:

        predictions = model(X_batch)
        loss = loss_function(predictions, y_batch)

        optimizer.zero_grad()
        loss.backward()
        optimizer.step()

        total_train_loss += loss.item()

    avg_train_loss = total_train_loss / len(train_loader)

    # Validation
    model.eval()

    with torch.no_grad():
        val_predictions = model(X_val_tensor)
        val_loss = loss_function(
            val_predictions, y_val_tensor
        ).item()

    # Check for improvement
    if val_loss < best_val_loss:
        best_val_loss = val_loss
        best_weights = copy.deepcopy(model.state_dict())
        epochs_without_improvement = 0
    else:
        epochs_without_improvement += 1

    if (epoch + 1) % 10 == 0:
        print(
            f"Epoch {epoch+1}: "
            f"Train Loss={avg_train_loss:.2f}, "
            f"Val Loss={val_loss:.2f}"
        )

    if epochs_without_improvement >= patience:
        print(f"\nEarly stopping at epoch {epoch+1}")
        break

# Restore best weights
model.load_state_dict(best_weights)

print("\nBest validation loss:", best_val_loss)
```

```python
class EnergyNN_v3(nn.Module):
    def __init__(self):
        super().__init__()

        self.network = nn.Sequential(
            nn.Linear(11, 128),
            nn.ReLU(),
            nn.Dropout(0.2),

            nn.Linear(128, 64),
            nn.ReLU(),
            nn.Dropout(0.2),

            nn.Linear(64, 32),
            nn.ReLU(),

            nn.Linear(32, 1)
        )

    def forward(self, x):
        return self.network(x)
```

```python
import copy

model_v3 = EnergyNN_v3()

loss_function = nn.MSELoss()

optimizer_v3 = torch.optim.Adam(
    model_v3.parameters(),
    lr=0.001
)

epochs = 150
patience = 15

best_val_loss_v3 = float("inf")
best_weights_v3 = None
epochs_without_improvement = 0

for epoch in range(epochs):

    model_v3.train()
    total_loss = 0

    for X_batch, y_batch in train_loader:

        predictions = model_v3(X_batch)
        loss = loss_function(predictions, y_batch)

        optimizer_v3.zero_grad()
        loss.backward()
        optimizer_v3.step()

        total_loss += loss.item()

    average_loss = total_loss / len(train_loader)

    model_v3.eval()

    with torch.no_grad():
        val_predictions = model_v3(X_val_tensor)
        val_loss = loss_function(
            val_predictions, y_val_tensor
        ).item()

    if val_loss < best_val_loss_v3:
        best_val_loss_v3 = val_loss
        best_weights_v3 = copy.deepcopy(
            model_v3.state_dict()
        )
        epochs_without_improvement = 0
    else:
        epochs_without_improvement += 1

    if (epoch + 1) % 10 == 0:
        print(
            f"Epoch {epoch+1}: "
            f"Train Loss={average_loss:.2f}, "
            f"Val Loss={val_loss:.2f}"
        )

    if epochs_without_improvement >= patience:
        print(f"Early stopping at epoch {epoch+1}")
        break

model_v3.load_state_dict(best_weights_v3)

print("Best V3 validation MSE:", best_val_loss_v3)
```

```python
model_v3.eval()

with torch.no_grad():
    val_predictions_v3 = model_v3(X_val_tensor).numpy()

actual_values = y_val_tensor.numpy()

val_mae_v3 = mean_absolute_error(
    actual_values, val_predictions_v3
)

val_rmse_v3 = mean_squared_error(
    actual_values, val_predictions_v3
) ** 0.5

print("V3 Validation MAE:", round(val_mae_v3, 4))
print("V3 Validation RMSE:", round(val_rmse_v3, 4))
```

```python
model_v4 = EnergyNN_v3()

loss_function = nn.MSELoss()

optimizer_v4 = torch.optim.Adam(
    model_v4.parameters(),
    lr=0.001
)
```

```python
scheduler = torch.optim.lr_scheduler.ReduceLROnPlateau(
    optimizer_v4,
    mode="min",
    factor=0.5,
    patience=5
)
```

```python
epochs = 150
early_stopping_patience = 15

best_val_loss_v4 = float("inf")
best_weights_v4 = None
epochs_without_improvement = 0

for epoch in range(epochs):

    model_v4.train()
    total_loss = 0

    for X_batch, y_batch in train_loader:

        predictions = model_v4(X_batch)
        loss = loss_function(predictions, y_batch)

        optimizer_v4.zero_grad()
        loss.backward()
        optimizer_v4.step()

        total_loss += loss.item()

    average_loss = total_loss / len(train_loader)

    model_v4.eval()

    with torch.no_grad():
        val_predictions = model_v4(X_val_tensor)
        val_loss = loss_function(
            val_predictions, y_val_tensor
        ).item()

    # Adjust learning rate based on validation loss
    scheduler.step(val_loss)

    if val_loss < best_val_loss_v4:
        best_val_loss_v4 = val_loss
        best_weights_v4 = copy.deepcopy(
            model_v4.state_dict()
        )
        epochs_without_improvement = 0
    else:
        epochs_without_improvement += 1

    if (epoch + 1) % 10 == 0:
        current_lr = optimizer_v4.param_groups[0]["lr"]

        print(
            f"Epoch {epoch+1}: "
            f"Train Loss={average_loss:.2f}, "
            f"Val Loss={val_loss:.2f}, "
            f"LR={current_lr}"
        )

    if epochs_without_improvement >= early_stopping_patience:
        print(f"Early stopping at epoch {epoch+1}")
        break

model_v4.load_state_dict(best_weights_v4)

print("Best V4 validation MSE:", best_val_loss_v4)
```

```python
model_v4.eval()

with torch.no_grad():
    val_predictions_v4 = model_v4(X_val_tensor).numpy()

actual_values = y_val_tensor.numpy()

val_mae_v4 = mean_absolute_error(
    actual_values, val_predictions_v4
)

val_rmse_v4 = mean_squared_error(
    actual_values, val_predictions_v4
) ** 0.5

print("V4 Validation MAE:", round(val_mae_v4, 4))
print("V4 Validation RMSE:", round(val_rmse_v4, 4))
```
